Create a bank account class

```python

class Account:
  def __init__(self, account_number: str, account_holder: str, account_balance: float = 0.0):
    self.account_number = account_number
    self.account_holder = account_holder
    self.account_balance = account_balance
    print(f"Account {self.account_number} created for {self.account_holder} with account_balance ${self.account_balance:.2f}.")

  def deposit(self, amount: float): # First method to add the amount passed as an argument to the account balance
    if amount > 0:
      self.account_balance += amount
      print(f"Deposited amount ${amount:.2f}. New balance: ${self.account_balance:.2f}.")
    else:
      print("Deposit amount must be positive.")

  def withdraw(self, amount: float): #second method; withdraw amount
    if amount <= 0:
      print("Withdrawal amount must be positive.")
    elif self.account_balance >= amount:
      self.account_balance -= amount
      print(f"Withdrew amount ${amount:.2f}. New balance: ${self.account_balance:.2f}.")
    else:
      print(f"Insufficient funds. Current balance: ${self.account_balance:.2f}. Attempted withdrawal: ${amount:.2f}.")

  def check_balance(self) -> float: #third method checks balance
    print(f"Current balance for account {self.account_number}: ${self.account_balance:.2f}.")
    return self.account_balance

# create an instance of the Account class and assign it to a variable called my_account
my_account = Account("78893456", "Alice Smith", 100.00)

# use methods to deposit money, withdraw money and check account balance
my_account.deposit(500.00)
my_account.withdraw(150.00)
my_account.check_balance()
my_account.withdraw(700.00) #attempt to withdraw more money than available
print("\n-------\n")

# create another account instance
bob_account = Account("987654321", "Bob Johnson", 200.00)

#use methods to deposit money, withdraw money and check account balance
bob_account.deposit(200.00)
bob_account.withdraw(50.00)
bob_account.check_balance()
print("\n-------\n")

# create Grace's account
Grace_account = Account("112233445", "Grace White", 1000.00)
Grace_account.deposit(600.00)
Grace_account.withdraw(150.00)
Grace_account.check_balance()


```
